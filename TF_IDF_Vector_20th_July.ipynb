# TF-IDF Vectorizer:-
## TF-IDF:-(Term Frequency-Inverse Document Frequency):-
### TF(Term Frequency):-
<i>
<pre>
    Term frequency-inverse document frequency gives a measure that takes the importance consideration depending on how frequently it occurs in a document and a corpus.
     To understand TF-IDF,Firstly we will understand the two terms separately.
     1. TF(Term Frequency)
     2. IDF(Inverse Document Frequency)
</pre>
</i> 

### TF(Term Frequency):-
<pre>
   <i> It is defined as the no. of time a word appear in a document divided by total no of words in the document.</i>
</pre>

### IDF(Inverse Document Frequency):-
<pre>
    <i>It is defined as the log of total no. of documents divided by no. of documents that contains the words.<i/>
</pre>
<i>IDF = log(N/DF(i))</i>
<pre>
    i = words 
    N = total no. of documents
    DF(i) = No. of documents that contain the word
</pre>

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()

txt = ["I love Singing and I love painting","I love ice-cream"]
z = tfidf.fit_transform(txt)
z
```

```python
print(z)
```

```python
print(tfidf.vocabulary_)
```
