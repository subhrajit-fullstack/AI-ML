```python
# Concatenation
a = np.random.randint(1,20,(2,2))
a
```

```python
b = np.random.randint(1,20,(2,4))
b
```

```python
c = np.random.randint(1,20,(4,4))
c
```

```python
np.concatenate((a,b))
```

```python
np.concatenate((a,b),axis=1)
```

```python
# axis=0 (row wise) (by default)
np.concatenate((b,c))
```

```python
# Stack:-
# Horizontal Stacking (axis=1) -> Row Equal
# Vertical Stacking   (axis=0) -> Column Equal
# Depth Stacking               -> Both must be equal
```

```python
np.hstack((a,b))
```

```python
np.vstack((b,c))
```
