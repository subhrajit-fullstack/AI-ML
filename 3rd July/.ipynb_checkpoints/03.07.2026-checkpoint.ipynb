### 5)Identity Operator('is', 'is not'):-

```python
a = 5
b = 5 
a is b
```

```python
a is not b
```

```python
c = 300
d = 300
c is d
```

```python
print(id(a))
print(id(b))
```

```python
print(id(c))
print(id(d))
```
