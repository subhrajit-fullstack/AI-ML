## Filtering On Image:-

```python
plt.imshow(img)
plt.show()
```

```python
rgb_img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(rgb_img)
plt.show()
```

```python
gray = cv2.cvtColor(rgb_img,cv2.COLOR_RGB2GRAY)
print(gray.shape)
plt.imshow(gray, cmap = 'gray')
plt.show()
```

### Gaussian blur:-

```python
blur = cv2.GaussianBlur(rgb_img,(303,33),20000) # (33,33):-Its called kernel matrix
plt.imshow(blur)
plt.show()
```

```python
# Canny edge detection:-
edge = cv2.Canny(rgb_img,50,120)
plt.imshow(edge,cmap = 'gray')
plt.show()
```

## Face detection :-

```python
img = cv2.imread(r'/Users/nexusop03/Downloads/shraddha.jpeg')
plt.imshow(img[::,::,::-1])
plt.grid()
plt.show()
```

```python
cv2.rectangle(img,(220,50),(480,350),(0,255,0),4)
plt.imshow(img[::,::,::-1])
plt.show()
```

```python
face = cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
model = cv2.CascadeClassifier(face)
img = cv2.imread(r"/Users/nexusop03/Downloads/shraddha.jpeg")
facepoints = model.detectMultiScale(img)
for x,y,w,h in facepoints:
    cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),4)
plt.imshow(img[::,::,::-1])
plt.show()
```

```python
face = cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
cap = cv2.VideoCapture(0)
ret, frame = cap.read()

while ret:
    ret, frame = cap.read()
    if not ret :
        break
    model = cv2.CascadeClassifier(face)
    facepoints = model.detectMultiScale(frame)
    for x,y,w,h in facepoints:
        cv2.rectangle(frame, (x,y),(x+w,y+h),(0,255,0),3)
    cv2.imshow('My Photo',frame)
    if cv2.waitKey(5) == ord('c'):
        break
cv2.destroyAllWindows()
cap.release()
```
