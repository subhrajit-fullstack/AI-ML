```python
import matplotlib.pyplot as plt
```

```python
x = [2,4,5,7,8,5,3]
y = [4,3,5,6,2,3,9]
plt.plot(x, y, color='c', linewidth=2, marker='o')
plt.title('Line color:',color = 'r')
plt.xlabel('X - Axis',color = 'b')
plt.ylabel('Y - Axis',color = 'g')
plt.show()
```

```python
x = [2,4,5,7,8,5,3]
y = [4,3,5,6,2,3,9]
plt.plot(x, y, 'ob-.') # in place of o we can use p,P,^ and etc...
plt.title('Line color:',color = 'r')
plt.xlabel('X - Axis',color = 'b')
plt.ylabel('Y - Axis',color = 'g')
plt.xticks(rotation = 45)
plt.grid()
plt.show()
```

```python
x = [2,4,5,7,8,5,3]
y = [4,3,5,6,2,3,9]
plt.plot(x, y, 'b-.') 
plt.plot(x,y , 'om')
plt.title('Line color:',color = 'r')
plt.xlabel('X - Axis',color = 'b')
plt.ylabel('Y - Axis',color = 'g')
plt.xticks(rotation = 45)
plt.grid()
plt.show()
```

```python
x = [2,4,5,7,8,5,3]
y = [4,3,5,6,2,3,9]
plt.bar(x, y)
plt.show()
```

```python
x = [0,2,4,6,8]
y = [1,3,5,7,9]
plt.bar(x, y, color = ['b', 'g'],width =0.9) # width generally 0.8 hota hai
plt.title('Bar plot',color = 'r')
plt.annotate('2026',xy = (3.5, 9))
plt.show()
```

```python
x = [0,2,4,6,8]
y = [1,3,5,7,9]
plt.bar(x, y ,color = 'g', width = 0.8, label ="Boys")
x = [1,3,5,7]
y = [2,4,6,8]
plt.bar(x, y ,color ='c', label ="Girls")
plt.title('Bar plot',color = 'y')
plt.annotate("2026",xy = (1.6,3.2))
plt.legend()
plt.show()
```
