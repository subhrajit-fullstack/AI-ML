<pre>

    1) Linear :- Continuous data jab hota hai
    2) Sigmoid :- Categorical Data ke liye use hota hai

    4) Tan H :- Categorical Data ke liye use hota hai

### Single Layer Perceptron:-

```python
import tensorflow
import keras
```

```python
from keras.models import Sequential # Layers ko sequentially rakhta hai 
from keras.layers import Dense # Layer banake ka kaam karta hai
from keras.optimizers import Adam # Model ko underfitting and overfitting hone nahi deta
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([3,6,8,6,9,5,8,3])
y = np.array([8,2,4,7,5,6,4,5])
```

```python
# Single Layered Perceptron
model = Sequential()
# input layer and first hidden layer 
model.add(Dense(1,input_dim = 1,activation = 'linear'))
#output layer
model.add(Dense(1,activation='linear'))
model.compile(Adam(learning_rate=0.03),loss = 'mse')
```

```python
model.summary()
```

```python
model.fit(x,y,epochs=20,verbose=1)
```

## Multi layered Perceptron:-

```python
# importing packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Reading and Studying Dataset
adv = pd.read_csv(r"/Users/nexusop03/Downloads/advertising.csv")
adv
```

### i/o creation:-

```python
ip = adv.drop('Sales',axis = 1)
op = adv.Sales
```

### train test split:-

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(ip,op,test_size=0.2)
```

```python
# Standard Scalar Transform:-
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()

x_train = sc.fit_transform(x_train)
x_test = sc.fit_transform(x_test)
```

```python
model = Sequential()
model.add(Dense(30,input_dim = 3,activation = 'linear'))
model.add(Dense(25,activation = 'linear'))
model.add(Dense(10,activation = 'linear'))
model.add(Dense(1,activation = 'linear'))
model.compile(Adam(learning_rate=0.03),loss = 'mse')
```

```python
model.summary()
```

```python
model.fit(x_train,y_train,validation_data = (x_test,y_test),epochs = 20)
```

```python
pred = model.predict(x_test)
pred
```

```python
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y_test,pred)
r2 = r2_score(y_test,pred)

print("MSE:",mse)
print("r2_score:",r2)
```
