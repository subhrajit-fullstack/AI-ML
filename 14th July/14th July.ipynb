```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
c=adv.corr()
c
```

# Input-Output Creation

```python
ip = adv[['TV','Radio','Newspaper']]  #(Another Method)
```

```python
ip = adv.drop('Sales',axis=1)
op = adv.Sales
```

```python
ip.head()
```

```python
op.head()
```

# Train-Test-Split

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test= train_test_split(ip,op,test_size=0.2,random_state=3000)
```

```python
x_train
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

# Standard Scalar

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

It replaces each value of x column with the new each value of Standard Scaler

```python
x_train  #range:- (-2,+2)
```

```python
from sklearn.preprocessing import MinMaxScaler
mm = MinMaxScaler()
x_train = mm.fit_transform(x_train)
x_test = mm.transform(x_test)
```

```python
x_train     # range:- (0,1)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
```

# Prediction

```python
pred = lr.predict(x_test)
pred
```

```python
y_test
```

## Average error kitna hai ye nikalna hoga, for that we use mean square error(mse).

# Accuracy

```python
from sklearn.metrics import mean_squared_error, r2_score
mse= mean_squared_error(y_test,pred)
r2 = r2_score(y_test,pred)

print("MSE : ",mse)
print("R2 Score : ",r2)
```

### To make the accuracy constant we use random_state, which we already use in the train-test above.

```python
# lmplot- Linear model plot
df = pd.DataFrame({'Y_Test':list(y_test), 'Prediction': pred})
df
```

```python
sns.lmplot(x='Y_Test',y='Prediction',data=df,scatter_kws={'color':'g'},line_kws={'color':'r'})
plt.show()
```
