## Basic Terminology used in NLP:-
<pre>
    1) Corpus - Large and Structured set of text used for linguistic research or NLP
    2) Document - A string from corpus stored in a list
    3) Word - A string from document 
    4) Vocabulary - Unique words in the document
    5) Vector - A string which have magnitude and direction(Magnitude means in a string / sentence which word has more focus)
    6) Stopwords - Commonly used words 
    7) Tokens
</pre>
### Steps involved in NLU/text preprocessing:-
<pre>
    1) Tokenization 
    2) Steaming 
    3) Lematization
    4) POS(Parts Of Speech) tags
    5) NER(Named Entity Recognition)
    6) Chunking
</pre>

### 1)Tokenization:-
<pre>breaking the big sentences into smaller units call tokens.
 <i>   It's two types:-
    1)sent_tokenize() - It's for sentence
    2)word_tokenize() - Its for word
     e.g. :- "God is Great" After tokenization It will be ['God','is','Great']
             
 </i>
</pre>

### 2)Steaming:-
<pre>
    It is used to normalize the words into their root / base form by removing suffixes.
    In this process it does not give perfect words thats why we use Lemmatization.
    e.g. :- studied = studi (which is not a proper word)
</pre>

### 3)Lemmatization:-
<pre>
    Converts words into their meaningful dictionary form is known as Lemma.Its is similar to steaming but O/P of lemmatization is a proper word.
    e.g. :- weighted = weight (It is a proper word)
            studied = study 
</pre>

### 4)POS tags:-
<pre>
    It is used to assigns grammatical roles to the words.
    e.g.:- The cat killed the rat.
</pre>

### 5)NER:-
<pre>
    It is a process of detecting important named entities.
    e.g.:- 'Search something on the <b>Google</b>.' - here google refers to a browser.
           'Sundar Pichai is the CEO of <b>Google</b>.' - here google refers to a organization.
    
</pre>

### 6)Chunking:-
<pre>
    Picking individiual piece of information and grouping them into a biggeer piece or phrase is called chunkimg.
    e.g. - The extra dark life of <b>ME</b>
</pre>

```python
# ne_chunk:-
txt = 'Narendra Modi is the Prime Minister of India'
print(txt)
```

```python
txt_token = word_tokenize(txt)
txt_token
```

```python
txt_tag = nltk.pos_tag(txt_token)
txt_tag
```

```python
from nltk import ne_chunk
txt_chunk = ne_chunk(txt_tag)
print(txt_chunk)
```

## Vectorizer:-
<pre>
    This technique used to convert text into numerical vectors so that machine learning algo can process text data.
    It is of 2 types:-
    1) count vectorizer
    2) TFIDF vectorizer
</pre>
### Count Vectorizer:-
<pre>
    It creates a vocabulary of all unique words and count how many time each word appear in a document.
    It converts all the character into lower case.
    It takes the unique words from the sentence and creates a sparse matrix.
    It removes all the stopwords from the sentence
    Arrange alphabatically
    
    Sparse Matrix:- 
</pre>

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
txt = ["I love singing and I love painting."]
x = cv.fit_transform(txt)
x
```

```python
print(x)
```

```python
print(cv.vocabulary_)
```

```python
print(x.toarray())
```

```python
txt = ["I love singing and I love painting.","I love ice ceram"]
txt
```

```python
y = cv.fit_transform(txt)
y
```

```python
print(y)
```

```python
print(cv.vocabulary_)
```

```python
print(y.toarray())
```
