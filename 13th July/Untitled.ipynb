# Co-relation

<pre>
    <b>
                      n       _        _
    Co-relation/Z/R = ∑ (x<sub>i</sub> - x) (y<sub>i</sub> - y)
                     i=1
                    ___________________________________________
                            n       _    n      _
                        √(  ∑ (x<sub>i</sub> - x)<sup>2</sup>  ∑ (y<sub>i</sub> - y)<sup>2</sup>)
                           i=1          i=1
    </b> 
</pre>

###delete karne ke liye drop function use hota hai

```python
iris.drop('class',axis = 1, inplace = True)
iris
```

```python
c = iris.corr()
c
```

```python
#Correlation ko plot karne ke liye heatmap use hota hai
sns.heatmap(c)
plt.show()
```

```python
sns.heatmap(c,annot = c)
plt.show()
```

```python
sns.heatmap(c,annot = c,cmap = 'Blues')
plt.show()
```
