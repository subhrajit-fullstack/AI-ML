```python
for i in 'Statement':
    if i == 'm':
        break
    print(i)
```

```python
## 2)Continue Statement:-
<pre><i>The continue keyword return control of the iteration to the beginning of the ython program of the python loop or 
python while loop.All remaining lines in the prevailing iteration of the loop are skipped by the continue keyword, which returns execution to the beginning of the next iteration of the loop.</i></pre>
```

```python
for i in 'Statement':
    if i == 'e':
        continue
    print(i)
```

```python
# Problem
def isPrime(a):
    fact=0
    for i in range(1,a+1):
        if a%i==0:
            fact+=1
    if fact==2:
        print(a,'is a prime number ')
    else:
        print(a, 'is not a prime number ')
isPrime(23)
```

```python
#understand to remember
def isPrime():
    n = int(input("Enter a number:"))
    c = 0
    for i in range(2, n):
        if n % i == 0:
            c += 1
    if c == 0:
        return 'Prime'
    else:
        return 'Not Prime'
isPrime()
```
