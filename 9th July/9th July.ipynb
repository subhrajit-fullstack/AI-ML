```python
df = pd.DataFrame({'Days':[1,2,3,4,5],
                  'Places':['Puri','Mathura','Asansol','Manali','Kolkata'],
                  'Visitors':[5895,5600,4859,5000,5800]})
df
```

```python
# to check missing values
df.isnull()
```

```python
df.isnull().sum()
```

```python
df = pd.DataFrame({'Days':[1,2,3,4,5],
                  'Places':['Puri','Mathura',np.nan,'Manali','Kolkata'],
                  'Visitors':[5895,5600,4859,5000,5800]})
df
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
import warnings 
warnings.filterwarnings('ignore')
```

```python
# fillna - to fill null values
df.Places.fillna('Puri',inplace=True)       # fillna is used only for nan
df
```
