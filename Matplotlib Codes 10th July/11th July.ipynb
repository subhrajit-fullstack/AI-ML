# Histogram:-

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ages = [12, 15, 17, 18, 20, 23, 21, 30, 33, 49, 48, 50]
bins = [10, 20, 30, 40, 50]

plt.hist(ages,bins, width = 0.8, edgecolor ='k', linewidth = 3, color = 'g')
plt.show()
# rwidth = bar ka width
# edge color = bar ka edge ka color
```

# ScatterPlot:-

```python
x = [2, 4, 3, 6, 9, 5]
y = [3, 4, 6, 7, 8, 9]

plt.scatter(x,y,s = 200, color = 'b',marker = 'o')
plt.show()
```

```python
x = [2, 4, 3, 6, 9, 5]
y = [3, 4, 6, 7, 8, 9]
plt.bar(x,y, width =0.08, color ='g')
plt.scatter(x,y,s = 200, color = 'g',marker = '*', edgecolor ='gold',linewidth = 10)
plt.scatter(x,y, s = 50, color ='brown', marker ='o')
plt.show()
```

# PieChart:-

```python
Kam = ['mobile','fresh','eat','study','travels','sleep']
slices = [8,2,3,3,2,6]
plt.pie(slices,labels = Kam, colors = ['r','b','gold','m','c','g'],
       startangle = 0, autopct = '%1.1f%%',shadow =True, explode = (0,0,0.09,0,0,0))
plt.show()
```
