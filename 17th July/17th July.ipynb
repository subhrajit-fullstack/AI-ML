## K Means Clustering


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
iris = pd.read_csv(r"/Users/nexusop03/Downloads/iris/iris.data",header=None)
iris.head()
```

```python
iris.columns = ['SL','SW','PL','PW','Flower']
iris.head(3)
```

```python
iris.isnull().sum()
```

```python
ip = iris.drop('Flower',axis=1)
ip.head(2)
```

```python
!pip uninstall scikit-learn
```

```python
from sklearn.cluster import KMeans
km = KMeans(n_clusters=3)
km.fit(ip)

km_pred = km.predict(ip)
km_pred
```

```python
centroid = km.cluster_centers_
centroid
```

```python
iris["Prediction"] = km_pred
iris
```

```python
sns.scatterplot(x=iris.SL,y=iris.PL,hue=iris['Flower'])
plt.show()
```

```python
sns.scatterplot(x=iris.SW,y=iris.PW,c=km_pred)
plt.show()
```

```python
sns.scatterplot(x=iris.SL,y=iris.PL,c=km_pred)
sns.scatterplot(x=centroid[::,0],y=centroid[::,2],s=100,color='k')
plt.show()
```
