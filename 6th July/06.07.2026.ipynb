```python
#Write a python program to check whether the entered number is prime or not.
a=int(input("Enter a number: "))
if a<=1:
    print("Not Prime ")
else:
    for i in range(2,a):
        if a%i==0:
            print("Not Prime ")
            break
    else:
        print("Prime ")
```

```python
a=int(input("Enter a number: "))
fact=0
for i in range(1,a+1):
    if a%i==0:
        fact+=1
if fact==2:
    print(a,'is a prime number ')
else:
    print(a, 'is not a prime number ')
```

```python
print('All the prime numbers between 1 to 100 are: ')
for a in range(1,101):
    fact=0
    for i in range(1,a+1):
        if a%i==0:
            fact+=1
    if fact==2:
        print(a,end=' ')
        
```

## while Loop:-

```python
# initialization
# while condition:
#       body of while 
#      increment/decrement
```

```python
a=int(input("Enter a number: "))
i=1
while i<=10:
    print(a,"x",i,"=",i*a)
    i+=1
```

```python
# By using while and nested while loop for prime numbers.
```

```python
n=int(input("Enter a number: "))
if n<=1:
    print("Not Prime ")
else:
    i=2
    while i<n:
        if n%i==0:
            print("Not Prime ")
            break
        i=i+1
    else:
        print("Prime ")
```

```python
n=int(input("Enter a number: "))
if n<=1:
    print("Not Prime ")
else:
    i=2
    while i<=2:
        j=2
        while j<n:
            if n%j==0:
                print("Not Prime ")
                break
            j=j+1

        if j==n:
            print("Prime ")
        i=i+1
```

```python
n=int(input("Enter the limit: "))
i=2
while i<=n:
    j=2
    while j<i:
        if i%j==0:
            break
        j=j+1
    if j==i:
        print(i,end=' ')
    i=i+1
```
